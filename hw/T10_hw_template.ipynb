**(b)에서 구한 한 그룹의 표준오차**를 기준으로, $0.02 \ge 3 \times \mathrm{SE}(n)$ 이 되는 가장 작은 $n$ 을 구한다.

### ✏️ TODO 4 — (d) 시뮬레이션: 올바른 결론에 얼마나 자주 도달하는가 · 2점

A/B 두 그룹을 **1,000번** 새로 만들어, **"B의 전환율이 A보다 높게 나온 비율"** 을 구한다.
한 그룹의 전환 수는 `rng.binomial(n, p)` 로 한 번에 뽑을 수 있다 (Topic 4 Binomial).

```python
def prob_B_wins(n, trials=1000):
    conv_A = rng.binomial(n, p_A, size=trials) / n     # 1,000번 실험의 A 전환율
    # TODO 4: B 그룹의 전환율 conv_B 를 같은 방식으로 만들고, conv_B > conv_A 인 비율을 돌려주세요
    conv_B = conv_A
    return 0.0


win_full = prob_B_wins(n_need)
print(f"n = {n_need:,}  ->  B 가 A 보다 높게 나온 비율 = {win_full:.3f}")
```

## 문제 2 — CLT는 언제 실패하는가 (10점)

| 분포 | 생성 코드 | 분산 |
|---|---|---|
| A | `rng.exponential(1, size)` | 유한 (1) |
| B | `rng.pareto(a=1.5, size=size)` | **무한대** |
| C | `rng.pareto(a=3.0, size=size)` | 유한 |

### ✏️ TODO 5 — (a) 3×3 히스토그램 · 4점

각 분포에서 $n = 10, 100, 1000$ 개씩 뽑아 평균을 낸다. 이를 **3,000번** 반복해 표본평균의 히스토그램을 그린다.

```python
dists = {
    "A: Exponential(1)": lambda size: rng.exponential(1, size),
    "B: Pareto(a=1.5)": lambda size: rng.pareto(1.5, size),
    "C: Pareto(a=3.0)": lambda size: rng.pareto(3.0, size),
}
NS = [10, 100, 1000]
REPEATS = 3000

means = {}                                     # means[(분포 이름, n)] = 표본평균 3,000개
for name, sampler in dists.items():
    for n in NS:
        # TODO 5: (REPEATS, n) 모양으로 뽑아 행마다 평균을 내세요
        #         힌트 - sampler((REPEATS, n)).mean(axis=1)
        means[(name, n)] = np.zeros(REPEATS)

fig, axes = plt.subplots(3, 3, figsize=(12, 8.5))
for i, name in enumerate(dists):
    for j, n in enumerate(NS):
        m = means[(name, n)]
        ax = axes[i, j]
        if m.std() > 0:
            lo, hi = np.percentile(m, [0.5, 99.5])
            ax.hist(m, bins=60, range=(lo, hi), color="#3b4fd8", alpha=0.8)
        ax.set_title(f"{name}, n={n}", fontsize=10)
        ax.set_yticks([])
plt.tight_layout()
plt.show()
```

### ✏️ TODO 6 — (b) 표본평균의 표준편차 표 · 2점

$n$ 이 10배 커질 때 표준편차가 $1/\sqrt{10} \approx 0.316$ 배로 줄어드는지 비율 열로 확인한다.

```python
print(f"{'distribution':<20}{'n=10':>10}{'n=100':>10}{'n=1000':>10}")
for name in dists:
    # TODO 6: 세 n 에 대한 표본평균의 표준편차를 구하고, 비율 sds[1]/sds[0], sds[2]/sds[1] 도 출력하세요
    sds = [0.0, 0.0, 0.0]
    print(f"{name:<20}{sds[0]:10.4f}{sds[1]:10.4f}{sds[2]:10.4f}")
print("이론 비율 1/sqrt(10) =", round(1 / np.sqrt(10), 3))
```
