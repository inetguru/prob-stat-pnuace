### ✏️ TODO 1 — Bayes 한 단계를 함수로 (b) 4점

양성이 한 번 나왔을 때 prior 를 posterior 로 갱신하는 함수다.

$$\text{posterior} = \frac{\text{sens} \times \text{prior}}
{\text{sens} \times \text{prior} + (1-\text{spec}) \times (1-\text{prior})}$$

분모가 **전확률 법칙**이다 — 진짜 양성과 오탐을 더한 것.

```python
def update(prior, sens=SENS, spec=SPEC):
    """양성이 한 번 더 나왔을 때 prior 를 posterior 로 갱신한다."""
    # TODO 1: 위 식을 코드로 옮기세요
    #         힌트 - 분자는 sens * prior,
    #                분모는 그 값 + (1 - spec) * (1 - prior)
    return prior        # <- 이 줄을 고치세요


first = update(PREVALENCE)
second = update(first)

print(f"1회 양성 후 : {first:.6f}  ({first:.2%})")
print(f"2회 양성 후 : {second:.6f}  ({second:.2%})")
```

> 💡 **두 검사는 조건부 독립**이라고 가정한다 — 환자 여부가 주어지면 두 결과가 서로 독립이다.
> 그래서 첫 번째 posterior 를 두 번째의 **prior 로 그대로 넣으면** 된다.
>

### ✏️ TODO 2 — $k$번 연속 양성일 때의 곡선 (c) 3점

$k = 1, 2, \ldots, 6$ 에 대해 posterior 를 차례로 계산해 리스트에 담는다.
**앞의 결과를 다음의 prior 로** 넘기는 것이 핵심이다.

```python
K = np.arange(1, 7)

post = []
cur = PREVALENCE
for k in K:
    # TODO 2: cur 를 한 단계 갱신하고 post 에 담으세요
    #         힌트 - cur = update(cur)  그다음  post.append(cur)
    post.append(cur)        # <- 갱신을 빠뜨리면 계속 0.001 이 나온다

post = np.array(post)

for k, v in zip(K, post):
    print(f"  k = {k} : {v:.6f}")
```

### ✏️ TODO 3 — 곡선을 그리고 90% 를 넘는 지점 표시 (c 계속)

> 📌 라벨은 **영어**로. Colab 에는 한글 폰트가 없어 글자가 □□□ 로 깨진다.

```python
plt.figure(figsize=(7, 4))
# TODO 3: post 곡선을 그리세요
#         힌트 - plt.plot(K, post, "o-", lw=2, label="P[ill | k positives]")

# 90% 를 처음 넘는 k 를 구한다 (post 가 비어 있지 않으면)
cross = int(K[np.argmax(post > 0.9)]) if (post > 0.9).any() else 0

plt.axhline(0.9, color="gray", ls="--", lw=1)
if cross:
    plt.axvline(cross, color="red", ls="--", lw=1.2)
    plt.annotate(f"k = {cross}", xy=(0.35, 0.45), xycoords="axes fraction",
                 color="red", fontweight="bold")

plt.xlabel("Number of consecutive positive tests (k)")
plt.ylabel("P[ill | k positives]")
plt.title("Repeated testing updates the prior")
plt.ylim(0, 1.05)
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print(f"90%를 처음 넘는 k = {cross}")
```

## 문제 2 — 유병률이 다르면 (5점)

검사 성능은 그대로 두고 **유병률만** $0.01\%$ 부터 $10\%$ 까지 바꿔가며,
**한 번 양성일 때의 posterior** 를 그린다.

- x축은 **로그 스케일**로 (`plt.xscale("log")`)
- $y = 0.5$ 기준선을 긋고, posterior 가 50% 를 넘는 **임계 유병률**을 읽는다

### ✏️ TODO 4 — 곡선 계산

```python
prev = np.logspace(-4, -1, 200)      # 0.01% ~ 10%

# TODO 4: 각 유병률에 대한 posterior 를 구하세요
#         힌트 - update 는 배열을 그대로 받는다.  curve = update(prev)
curve = np.zeros_like(prev)          # <- 이 줄을 고치세요

plt.figure(figsize=(7, 4))
plt.plot(prev * 100, curve, lw=2, label="P[ill | one positive]")
plt.axhline(0.5, color="gray", ls="--", lw=1)

idx = np.argmax(curve > 0.5)
thr = prev[idx] * 100 if (curve > 0.5).any() else 0.0
if thr:
    plt.axvline(thr, color="red", ls="--", lw=1.2)

plt.xscale("log")
plt.xlabel("Prevalence (%)")
plt.ylabel("P[ill | one positive]")
plt.title("The same test, different populations")
plt.ylim(0, 1.05)
plt.legend()
plt.grid(alpha=0.3, which="both")
plt.show()

print(f"posterior가 50%를 넘는 유병률 ≈ {thr:.3f}%")
```

> 💡 검사기는 **하나도 안 바뀌었다.** 바뀐 것은 **검사받는 집단**이다.
> 증상이 있어서 병원에 온 사람은 prior 가 높고, 건강검진으로 온 사람은 낮다.
