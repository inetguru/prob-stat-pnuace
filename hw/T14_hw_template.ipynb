## 문제 1 — 검정력 실험 (8점)

두 그룹의 평균 차이가 $d$ (표준편차 1 기준, 즉 효과크기)일 때 검정력을 시뮬레이션으로 구한다.

### ✏️ TODO 1 — (a) 검정력을 구하는 함수 · 2점

`reps` 번의 실험을 **배열 하나로 한꺼번에** 만든다. 행 하나가 실험 하나다.
`stats.ttest_ind(a, b, axis=1)` 은 행마다 따로 검정해 p-value 배열을 돌려준다.

```python
def power(d, n, reps=2000, rng=rng):
    a = rng.normal(0, 1, size=(reps, n))     # 그룹 A: 평균 0
    b = rng.normal(d, 1, size=(reps, n))     # 그룹 B: 평균 d  (진짜 효과가 있다)
    # TODO 1: 행마다 t-test 를 해서 p < ALPHA 인 비율을 돌려주세요
    #         힌트: stats.ttest_ind(a, b, axis=1).pvalue
    return 0.0


print(f"d = 0.5, n = 50 의 검정력: {power(0.5, 50):.3f}")
```

### ✏️ TODO 2 — (b) 표와 그래프로 정리 · 2점

$d = 0.2, 0.5, 0.8$ 과 $n = 20, 50, 100, 200, 500$ 의 모든 조합에서 검정력을 구한다.
`POWER[i, j]` 에 $d$ 는 `D_LIST[i]`, $n$ 은 `N_LIST[j]` 인 검정력을 넣는다.

```python
D_LIST = [0.2, 0.5, 0.8]
N_LIST = [20, 50, 100, 200, 500]

POWER = np.zeros((len(D_LIST), len(N_LIST)))
for i, d in enumerate(D_LIST):
    for j, n in enumerate(N_LIST):
        # TODO 2: 이 조합의 검정력을 POWER[i, j] 에 넣으세요.  힌트: power(d, n)
        pass

print(f"{'d / n':>8}" + "".join(f"{n:>8}" for n in N_LIST))
for i, d in enumerate(D_LIST):
    print(f"{d:>8}" + "".join(f"{v:>8.3f}" for v in POWER[i]))

plt.figure(figsize=(7, 4))
for i, d in enumerate(D_LIST):
    plt.plot(N_LIST, POWER[i], "o-", lw=2, label=f"d = {d}")
plt.axhline(0.8, color="gray", ls="--", label="power 0.8")
plt.xscale("log")
plt.xticks(N_LIST, [str(n) for n in N_LIST])
plt.minorticks_off()
plt.ylim(0, 1.02)
plt.xlabel("n per group")
plt.ylabel("power")
plt.title("Power of a two-sample t-test (alpha = 0.05)")
plt.legend()
plt.show()
```

### ✏️ TODO 3 — (c) 검정력 0.8 에 필요한 $n$ · 2점

표의 $n$ 사이는 **직선으로 이어 읽는다**(선형 보간). `np.interp(0.8, 검정력들, n들)` 이 그 일을 한다.
검정력이 $n$ 과 함께 커지므로 검정력을 x 자리에 넣을 수 있다.
⚠️ 표 안에서 0.8 에 닿지 않으면 "500 넘게 필요하다"고 쓴다.

```python
for i, d in enumerate(D_LIST):
    if POWER[i, -1] < 0.8:
        print(f"d = {d}: n = 500 으로도 검정력 0.8 에 못 미친다 -> 500 넘게 필요")
    elif POWER[i, 0] >= 0.8:
        print(f"d = {d}: n = 20 이미 0.8 이상")
    else:
        # TODO 3: 검정력 0.8 에 해당하는 n 을 선형 보간으로 구하세요
        #         힌트: np.interp(0.8, POWER[i], N_LIST)
        need = 0.0
        print(f"d = {d}: 약 n = {need:.0f} 명 (각 그룹)")
```

## 문제 2 — 다중비교를 직접 겪기 (6점)

효과가 **0인** 데이터에서 **지표 15개**를 각각 검정하고, 그중 **가장 작은 p-value** 만 보고한다.
이 실험을 1,000번 반복한다.

### ✏️ TODO 4 — (a) 실험마다 가장 작은 p-value · 2점

배열 모양은 (실험 1,000 × 지표 15 × 사람 30) 이다. 마지막 축(`axis=2`)으로 검정하면
(1,000 × 15) 개의 p-value가 나온다.

```python
REPS, K, N_EACH = 1000, 15, 30
a = rng.normal(0, 1, size=(REPS, K, N_EACH))
b = rng.normal(0, 1, size=(REPS, K, N_EACH))      # 진짜 차이는 0
pvals = stats.ttest_ind(a, b, axis=2).pvalue      # 모양 (1000, 15)

# TODO 4: 실험(행)마다 15개 p-value 중 가장 작은 값을 구하세요.  힌트: pvals.min(axis=1)
min_p = np.ones(REPS)
print(f"min_p 개수 {len(min_p)} · 처음 다섯 개 {np.round(min_p[:5], 4)}")
```

### (b) "가장 작은 p-value" 의 분포 · 2점

아래 셀이 히스토그램을 그린다. 비교용으로 **지표 하나만** 본 경우(`pvals[:, 0]`)도 함께 그린다.

```python
plt.figure(figsize=(7, 4))
plt.hist(pvals[:, 0], bins=20, range=(0, 1), alpha=0.6, label="one metric (honest)")
plt.hist(min_p, bins=20, range=(0, 1), alpha=0.6, label="smallest of 15 metrics")
plt.axvline(ALPHA, color="red", ls="--", lw=2)
plt.xlabel("p-value")
plt.ylabel("count")
plt.title("Reporting only the smallest p-value")
plt.legend()
plt.show()
```
